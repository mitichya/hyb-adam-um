```python
from pathlib import Path
import re
from textwrap import dedent


# =========================
# 1. Paste your two tables here
# =========================

TABLE_CERCOPITHECIDAE = r"""





% Requires: \usepackage{booktabs,multirow,rotating}
% Recommended macro: \newcommand{\best}[1]{\textbf{#1}}
\begin{sidewaystable}[t]
\caption{Combined matrix-, topology-, and ML-branch-length performance on $15\times 15$ mtDNA distance matrices. Matrix errors are computed only on artificially hidden entries. RMSE/MAE, NJ-vs-ML patristic RMSE, and branch-length RMSE are reported as $\times 10^{-2}$. Lower is better for RMSE/MAE, runtime, RF$_{\mathrm{norm}}$, NJ-vs-ML patristic RMSE, and branch-length RMSE; higher is better for correlations. Best values within each missingness level are in bold.}
\label{tab:combined_main_results}
\centering
\tiny
\setlength{\tabcolsep}{2pt}
\begin{tabular}{llcccccccccc}
\toprule
& & \multicolumn{5}{c}{Matrix metrics} & \multicolumn{5}{c}{Tree / branch-length metrics} \\
\cmidrule(lr){3-7}\cmidrule(lr){8-12}
$p$ & Method & RMSE & MAE & Pear. & Spear. & Time(s) & RF$_{\mathrm{norm}}$ & NJ pat.\ RMSE & NJ pat.\ Spear. & edge BL RMSE & Succ. \\
\midrule
\multirow{6}{*}{30\%} & Hyb-Adam-UM & 1.51 $\pm$ 0.82 & 0.59 $\pm$ 0.34 & 0.90 $\pm$ 0.09 & 0.85 $\pm$ 0.12 & 21.48 $\pm$ 0.18 & 0.29 $\pm$ 0.12 & \best{15.36 $\pm$ 0.03} & 0.92 $\pm$ 0.03 & 3.53 $\pm$ 0.20 & 30/30 \\
 & MW$^\star$-proj & \best{1.19 $\pm$ 0.83} & \best{0.45 $\pm$ 0.33} & \best{0.93 $\pm$ 0.08} & \best{0.91 $\pm$ 0.08} & 15.74 $\pm$ 1.64 & \best{0.26 $\pm$ 0.11} & 15.38 $\pm$ $<0.01$ & \best{0.93 $\pm$ 0.02} & \best{3.52 $\pm$ 0.20} & 30/30 \\
 & NJ$^\star$-proj & 2.13 $\pm$ 1.19 & 1.15 $\pm$ 0.86 & 0.82 $\pm$ 0.17 & 0.71 $\pm$ 0.26 & \best{0.01 $\pm$ 0.00} & 0.38 $\pm$ 0.17 & 15.43 $\pm$ 0.22 & 0.87 $\pm$ 0.07 & 3.59 $\pm$ 0.25 & 29/30 \\
 & LRMC & 5.09 $\pm$ 0.55 & 4.21 $\pm$ 0.51 & 0.60 $\pm$ 0.16 & 0.51 $\pm$ 0.21 & 0.03 $\pm$ 0.01 & 0.68 $\pm$ 0.14 & 16.79 $\pm$ 0.31 & 0.73 $\pm$ 0.08 & 3.87 $\pm$ 0.30 & 30/30 \\
 & KNN-impute & 3.14 $\pm$ 0.33 & 2.45 $\pm$ 0.28 & 0.87 $\pm$ 0.05 & 0.76 $\pm$ 0.12 & 0.13 $\pm$ 0.72 & 0.49 $\pm$ 0.13 & 16.09 $\pm$ 0.17 & 0.88 $\pm$ 0.05 & 3.79 $\pm$ 0.35 & 30/30 \\
 & MDS-SMACOF & 1.93 $\pm$ 0.45 & 1.38 $\pm$ 0.31 & 0.89 $\pm$ 0.06 & 0.81 $\pm$ 0.10 & 0.05 $\pm$ $<0.01$ & 0.46 $\pm$ 0.15 & 15.67 $\pm$ 0.10 & 0.91 $\pm$ 0.02 & 3.59 $\pm$ 0.16 & 30/30 \\
\midrule
\multirow{6}{*}{50\%} & Hyb-Adam-UM & 1.76 $\pm$ 0.98 & 0.73 $\pm$ 0.49 & 0.88 $\pm$ 0.11 & 0.84 $\pm$ 0.09 & 34.55 $\pm$ 0.27 & 0.39 $\pm$ 0.14 & 15.39 $\pm$ 0.04 & 0.89 $\pm$ 0.04 & 3.71 $\pm$ 0.28 & 30/30 \\
 & MW$^\star$-proj & \best{1.59 $\pm$ 0.86} & \best{0.64 $\pm$ 0.42} & \best{0.90 $\pm$ 0.08} & \best{0.88 $\pm$ 0.07} & 14.35 $\pm$ 2.11 & \best{0.39 $\pm$ 0.13} & 15.40 $\pm$ 0.04 & \best{0.91 $\pm$ 0.03} & \best{3.70 $\pm$ 0.27} & 30/30 \\
 & NJ$^\star$-proj & 2.53 $\pm$ 1.24 & 1.37 $\pm$ 1.04 & 0.78 $\pm$ 0.16 & 0.72 $\pm$ 0.22 & 0.01 $\pm$ 0.00 & 0.53 $\pm$ 0.16 & \best{15.37 $\pm$ 0.43} & 0.81 $\pm$ 0.12 & 3.82 $\pm$ 0.32 & 22/30 \\
 & LRMC & 5.89 $\pm$ 0.56 & 4.94 $\pm$ 0.51 & 0.46 $\pm$ 0.16 & 0.37 $\pm$ 0.17 & 0.04 $\pm$ 0.05 & 0.83 $\pm$ 0.10 & 18.08 $\pm$ 0.45 & 0.55 $\pm$ 0.15 & 4.30 $\pm$ 0.32 & 30/30 \\
 & KNN-impute & 3.56 $\pm$ 0.39 & 2.80 $\pm$ 0.38 & 0.81 $\pm$ 0.05 & 0.73 $\pm$ 0.11 & \best{$<0.01$ $\pm$ $<0.01$} & 0.52 $\pm$ 0.15 & 16.72 $\pm$ 0.30 & 0.82 $\pm$ 0.06 & 4.06 $\pm$ 0.31 & 30/30 \\
 & MDS-SMACOF & 2.32 $\pm$ 0.60 & 1.66 $\pm$ 0.34 & 0.84 $\pm$ 0.10 & 0.77 $\pm$ 0.08 & 0.03 $\pm$ 0.01 & 0.59 $\pm$ 0.14 & 15.94 $\pm$ 0.13 & 0.86 $\pm$ 0.04 & 3.82 $\pm$ 0.20 & 30/30 \\
\midrule
\multirow{6}{*}{65\%} & Hyb-Adam-UM & \best{2.60 $\pm$ 0.95} & \best{1.17 $\pm$ 0.53} & \best{0.76 $\pm$ 0.13} & \best{0.80 $\pm$ 0.06} & 44.84 $\pm$ 0.79 & \best{0.59 $\pm$ 0.14} & \best{15.45 $\pm$ 0.06} & \best{0.84 $\pm$ 0.04} & 4.00 $\pm$ 0.28 & 30/30 \\
 & MW$^\star$-proj & 2.66 $\pm$ 0.95 & 1.24 $\pm$ 0.56 & 0.74 $\pm$ 0.15 & 0.76 $\pm$ 0.11 & 14.96 $\pm$ 1.90 & 0.60 $\pm$ 0.16 & 15.47 $\pm$ 0.16 & 0.83 $\pm$ 0.07 & \best{4.00 $\pm$ 0.27} & 30/30 \\
 & NJ$^\star$-proj & 4.11 $\pm$ 1.03 & 2.66 $\pm$ 1.07 & 0.47 $\pm$ 0.22 & 0.51 $\pm$ 0.24 & 0.01 $\pm$ 0.00 & 0.84 $\pm$ 0.12 & 15.56 $\pm$ 0.44 & 0.64 $\pm$ 0.18 & 4.38 $\pm$ 0.28 & 8/30 \\
 & LRMC & 7.01 $\pm$ 0.65 & 5.98 $\pm$ 0.63 & 0.30 $\pm$ 0.14 & 0.26 $\pm$ 0.16 & 0.06 $\pm$ 0.05 & 0.94 $\pm$ 0.11 & 19.57 $\pm$ 0.60 & 0.37 $\pm$ 0.14 & 4.64 $\pm$ 0.32 & 30/30 \\
 & KNN-impute & 4.60 $\pm$ 0.45 & 3.66 $\pm$ 0.42 & 0.63 $\pm$ 0.10 & 0.58 $\pm$ 0.12 & \best{$<0.01$ $\pm$ $<0.01$} & 0.69 $\pm$ 0.14 & 17.61 $\pm$ 0.44 & 0.71 $\pm$ 0.10 & 4.36 $\pm$ 0.36 & 30/30 \\
 & MDS-SMACOF & 3.08 $\pm$ 0.61 & 2.22 $\pm$ 0.35 & 0.61 $\pm$ 0.20 & 0.61 $\pm$ 0.15 & 0.01 $\pm$ $<0.01$ & 0.72 $\pm$ 0.12 & 16.32 $\pm$ 0.21 & 0.77 $\pm$ 0.08 & 4.11 $\pm$ 0.22 & 30/30 \\
\midrule
\multirow{6}{*}{85\%} & Hyb-Adam-UM & 4.09 $\pm$ 0.73 & \best{2.37 $\pm$ 0.66} & \best{0.35 $\pm$ 0.22} & \best{0.35 $\pm$ 0.28} & 60.49 $\pm$ 1.17 & \best{0.84 $\pm$ 0.13} & \best{15.43 $\pm$ 0.31} & 0.43 $\pm$ 0.27 & 4.57 $\pm$ 0.23 & 30/30 \\
 & MW$^\star$-proj & 4.43 $\pm$ 0.73 & 2.80 $\pm$ 0.65 & 0.26 $\pm$ 0.19 & 0.33 $\pm$ 0.21 & 14.84 $\pm$ 5.00 & 0.89 $\pm$ 0.10 & 15.97 $\pm$ 0.67 & 0.39 $\pm$ 0.21 & 4.60 $\pm$ 0.30 & 30/30 \\
 & NJ$^\star$-proj & N/A & N/A & N/A & N/A & N/A & N/A & N/A & N/A & N/A & 0/30 \\
 & LRMC & 10.14 $\pm$ 0.92 & 9.05 $\pm$ 1.01 & 0.09 $\pm$ 0.14 & 0.09 $\pm$ 0.16 & 0.12 $\pm$ 0.07 & 0.98 $\pm$ 0.04 & 23.56 $\pm$ 0.91 & 0.14 $\pm$ 0.17 & 5.31 $\pm$ 0.28 & 30/30 \\
 & KNN-impute & 6.97 $\pm$ 0.47 & 5.89 $\pm$ 0.48 & 0.20 $\pm$ 0.13 & 0.23 $\pm$ 0.14 & $<0.01$ $\pm$ $<0.01$ & 0.95 $\pm$ 0.06 & 20.54 $\pm$ 0.64 & 0.29 $\pm$ 0.16 & 4.84 $\pm$ 0.31 & 30/30 \\
 & MDS-SMACOF & \best{3.96 $\pm$ 0.45} & 3.07 $\pm$ 0.42 & 0.24 $\pm$ 0.16 & 0.26 $\pm$ 0.14 & \best{$<0.01$ $\pm$ $<0.01$} & 0.86 $\pm$ 0.07 & 17.26 $\pm$ 0.80 & \best{0.48 $\pm$ 0.16} & \best{4.55 $\pm$ 0.21} & 30/30 \\
\bottomrule
\end{tabular}
\vspace{2pt}
\parbox{\textwidth}{\scriptsize Succ. is displayed as successful/evaluated runs in the 30-run protocol; when the denominator is absent from a source file, it is set to 30. The numerator is computed from n_success when available, otherwise from n_failed or finite summary metrics. NJ patristic RMSE/Spearman compare pairwise leaf-to-leaf patristic distances between the NJ tree reconstructed from the completed matrix and the reference ML tree. No fixed-ML-topology fitted-patristic values are used as fallback in these main columns. edge BL RMSE is taken from the ML-topology branch-length comparison. N/A indicates that the corresponding summary file or metric column was not available.}
\end{sidewaystable}




"""

TABLE_HETEROGENEOUS = r"""




% Requires: \usepackage{booktabs,multirow,rotating}
% Recommended macro: \newcommand{\best}[1]{\textbf{#1}}
\begin{sidewaystable}[t]
\caption{Combined matrix-, topology-, and ML-branch-length performance on $15\times 15$ mtDNA distance matrices. Matrix errors are computed only on artificially hidden entries. RMSE/MAE, NJ-vs-ML patristic RMSE, and branch-length RMSE are reported as $\times 10^{-2}$. Lower is better for RMSE/MAE, runtime, RF$_{\mathrm{norm}}$, NJ-vs-ML patristic RMSE, and branch-length RMSE; higher is better for correlations. Best values within each missingness level are in bold.}
\label{tab:combined_main_results}
\centering
\tiny
\setlength{\tabcolsep}{2pt}
\begin{tabular}{llcccccccccc}
\toprule
& & \multicolumn{5}{c}{Matrix metrics} & \multicolumn{5}{c}{Tree / branch-length metrics} \\
\cmidrule(lr){3-7}\cmidrule(lr){8-12}
$p$ & Method & RMSE & MAE & Pear. & Spear. & Time(s) & RF$_{\mathrm{norm}}$ & NJ pat.\ RMSE & NJ pat.\ Spear. & edge BL RMSE & Succ. \\
\midrule
\multirow{6}{*}{30\%} & Hyb-Adam-UM & 1.56 $\pm$ 0.78 & 0.80 $\pm$ 0.36 & 0.96 $\pm$ 0.03 & 0.94 $\pm$ 0.04 & 21.65 $\pm$ 0.22 & 0.13 $\pm$ 0.11 & 45.32 $\pm$ 0.06 & 0.97 $\pm$ $<0.01$ & \best{9.05 $\pm$ 0.14} & 30/30 \\
 & MW$^\star$-proj & \best{1.34 $\pm$ 0.76} & \best{0.70 $\pm$ 0.34} & \best{0.97 $\pm$ 0.02} & \best{0.97 $\pm$ 0.02} & 14.87 $\pm$ 1.30 & \best{0.11 $\pm$ 0.10} & 45.36 $\pm$ 0.05 & \best{0.98 $\pm$ $<0.01$} & 9.06 $\pm$ 0.11 & 30/30 \\
 & NJ$^\star$-proj & 2.18 $\pm$ 0.89 & 1.37 $\pm$ 0.62 & 0.92 $\pm$ 0.04 & 0.87 $\pm$ 0.07 & \best{0.01 $\pm$ 0.01} & 0.18 $\pm$ 0.12 & \best{45.27 $\pm$ 0.17} & 0.95 $\pm$ 0.03 & 9.06 $\pm$ 0.15 & 27/30 \\
 & LRMC & 10.60 $\pm$ 0.79 & 9.14 $\pm$ 0.82 & 0.40 $\pm$ 0.15 & 0.35 $\pm$ 0.14 & 0.06 $\pm$ 0.04 & 0.88 $\pm$ 0.13 & 48.60 $\pm$ 0.39 & 0.62 $\pm$ 0.07 & 10.00 $\pm$ 0.54 & 30/30 \\
 & KNN-impute & 6.36 $\pm$ 0.49 & 5.42 $\pm$ 0.54 & 0.81 $\pm$ 0.07 & 0.73 $\pm$ 0.08 & 0.03 $\pm$ 0.14 & 0.48 $\pm$ 0.14 & 46.92 $\pm$ 0.35 & 0.81 $\pm$ 0.08 & 9.89 $\pm$ 0.47 & 30/30 \\
 & MDS-SMACOF & 3.93 $\pm$ 0.94 & 3.13 $\pm$ 0.61 & 0.78 $\pm$ 0.13 & 0.72 $\pm$ 0.11 & 0.04 $\pm$ $<0.01$ & 0.24 $\pm$ 0.15 & 45.93 $\pm$ 0.16 & 0.90 $\pm$ 0.06 & 9.20 $\pm$ 0.12 & 30/30 \\
\midrule
\multirow{6}{*}{50\%} & Hyb-Adam-UM & 2.12 $\pm$ 0.98 & 1.12 $\pm$ 0.51 & 0.93 $\pm$ 0.06 & 0.90 $\pm$ 0.08 & 35.00 $\pm$ 0.28 & 0.33 $\pm$ 0.16 & 45.31 $\pm$ 0.10 & 0.95 $\pm$ 0.04 & 9.03 $\pm$ 0.17 & 30/30 \\
 & MW$^\star$-proj & \best{2.04 $\pm$ 1.00} & \best{1.09 $\pm$ 0.52} & \best{0.93 $\pm$ 0.06} & \best{0.93 $\pm$ 0.07} & 14.14 $\pm$ 2.00 & \best{0.29 $\pm$ 0.14} & 45.39 $\pm$ 0.12 & \best{0.95 $\pm$ 0.03} & 9.05 $\pm$ 0.17 & 30/30 \\
 & NJ$^\star$-proj & 2.59 $\pm$ 1.04 & 1.56 $\pm$ 0.73 & 0.89 $\pm$ 0.07 & 0.88 $\pm$ 0.08 & 0.01 $\pm$ 0.00 & 0.41 $\pm$ 0.19 & \best{45.15 $\pm$ 0.21} & 0.92 $\pm$ 0.05 & \best{8.90 $\pm$ 0.27} & 22/30 \\
 & LRMC & 11.46 $\pm$ 0.81 & 9.88 $\pm$ 0.74 & 0.31 $\pm$ 0.13 & 0.28 $\pm$ 0.15 & 0.08 $\pm$ 0.05 & 0.93 $\pm$ 0.09 & 50.57 $\pm$ 0.61 & 0.47 $\pm$ 0.10 & 10.54 $\pm$ 0.57 & 30/30 \\
 & KNN-impute & 6.72 $\pm$ 0.55 & 5.52 $\pm$ 0.46 & 0.75 $\pm$ 0.08 & 0.70 $\pm$ 0.10 & \best{$<0.01$ $\pm$ $<0.01$} & 0.60 $\pm$ 0.15 & 47.97 $\pm$ 0.50 & 0.77 $\pm$ 0.08 & 10.43 $\pm$ 0.48 & 30/30 \\
 & MDS-SMACOF & 4.29 $\pm$ 0.85 & 3.43 $\pm$ 0.58 & 0.72 $\pm$ 0.13 & 0.71 $\pm$ 0.10 & 0.02 $\pm$ $<0.01$ & 0.46 $\pm$ 0.16 & 46.35 $\pm$ 0.30 & 0.85 $\pm$ 0.06 & 9.31 $\pm$ 0.16 & 30/30 \\
\midrule
\multirow{6}{*}{65\%} & Hyb-Adam-UM & 3.18 $\pm$ 1.17 & \best{1.72 $\pm$ 0.68} & \best{0.83 $\pm$ 0.12} & 0.81 $\pm$ 0.14 & 46.29 $\pm$ 0.75 & 0.56 $\pm$ 0.16 & 45.22 $\pm$ 0.17 & 0.89 $\pm$ 0.10 & 9.07 $\pm$ 0.17 & 30/30 \\
 & MW$^\star$-proj & \best{3.14 $\pm$ 1.15} & 1.75 $\pm$ 0.71 & 0.83 $\pm$ 0.12 & \best{0.86 $\pm$ 0.11} & 14.43 $\pm$ 1.74 & \best{0.51 $\pm$ 0.17} & 45.45 $\pm$ 0.29 & \best{0.91 $\pm$ 0.07} & 9.12 $\pm$ 0.18 & 30/30 \\
 & NJ$^\star$-proj & 4.90 $\pm$ 1.93 & 3.38 $\pm$ 1.54 & 0.65 $\pm$ 0.22 & 0.65 $\pm$ 0.26 & 0.01 $\pm$ 0.00 & 0.70 $\pm$ 0.18 & \best{44.86 $\pm$ 0.62} & 0.75 $\pm$ 0.16 & \best{8.94 $\pm$ 0.38} & 11/30 \\
 & LRMC & 13.34 $\pm$ 0.95 & 11.58 $\pm$ 0.80 & 0.18 $\pm$ 0.10 & 0.16 $\pm$ 0.10 & 0.10 $\pm$ 0.06 & 0.95 $\pm$ 0.05 & 53.21 $\pm$ 0.74 & 0.29 $\pm$ 0.13 & 11.05 $\pm$ 0.59 & 30/30 \\
 & KNN-impute & 8.25 $\pm$ 0.58 & 6.74 $\pm$ 0.61 & 0.58 $\pm$ 0.10 & 0.54 $\pm$ 0.09 & \best{$<0.01$ $\pm$ $<0.01$} & 0.75 $\pm$ 0.17 & 49.53 $\pm$ 0.51 & 0.67 $\pm$ 0.09 & 10.76 $\pm$ 0.49 & 30/30 \\
 & MDS-SMACOF & 4.86 $\pm$ 0.69 & 3.89 $\pm$ 0.48 & 0.56 $\pm$ 0.17 & 0.62 $\pm$ 0.13 & 0.01 $\pm$ $<0.01$ & 0.69 $\pm$ 0.15 & 46.93 $\pm$ 0.43 & 0.75 $\pm$ 0.08 & 9.46 $\pm$ 0.18 & 30/30 \\
\midrule
\multirow{6}{*}{85\%} & Hyb-Adam-UM & \best{5.73 $\pm$ 1.04} & \best{3.65 $\pm$ 0.87} & \best{0.46 $\pm$ 0.19} & \best{0.41 $\pm$ 0.28} & 60.61 $\pm$ 0.58 & \best{0.90 $\pm$ 0.08} & \best{44.67 $\pm$ 0.49} & \best{0.49 $\pm$ 0.26} & \best{9.30 $\pm$ 0.21} & 30/30 \\
 & MW$^\star$-proj & 6.96 $\pm$ 1.28 & 4.53 $\pm$ 1.08 & 0.23 $\pm$ 0.21 & 0.32 $\pm$ 0.25 & 14.22 $\pm$ 4.77 & 0.93 $\pm$ 0.08 & 45.92 $\pm$ 1.58 & 0.39 $\pm$ 0.23 & 9.68 $\pm$ 0.63 & 30/30 \\
 & NJ$^\star$-proj & N/A & N/A & N/A & N/A & N/A & N/A & N/A & N/A & N/A & 0/30 \\
 & LRMC & 18.79 $\pm$ 1.54 & 16.93 $\pm$ 1.73 & 0.03 $\pm$ 0.12 & $<0.01$ $\pm$ 0.16 & 0.16 $\pm$ 0.07 & 0.99 $\pm$ 0.02 & 60.21 $\pm$ 1.61 & 0.08 $\pm$ 0.16 & 12.58 $\pm$ 0.50 & 30/30 \\
 & KNN-impute & 12.77 $\pm$ 0.74 & 10.81 $\pm$ 0.71 & 0.14 $\pm$ 0.10 & 0.14 $\pm$ 0.12 & $<0.01$ $\pm$ $<0.01$ & 0.96 $\pm$ 0.06 & 54.67 $\pm$ 0.91 & 0.23 $\pm$ 0.15 & 11.43 $\pm$ 0.59 & 30/30 \\
 & MDS-SMACOF & 6.03 $\pm$ 1.05 & 4.83 $\pm$ 0.80 & 0.05 $\pm$ 0.23 & 0.11 $\pm$ 0.25 & \best{$<0.01$ $\pm$ $<0.01$} & 0.93 $\pm$ 0.06 & 48.01 $\pm$ 1.53 & 0.39 $\pm$ 0.20 & 9.93 $\pm$ 0.35 & 30/30 \\
\bottomrule
\end{tabular}
\vspace{2pt}
\parbox{\textwidth}{\scriptsize Succ. is displayed as successful/evaluated runs in the 30-run protocol; when the denominator is absent from a source file, it is set to 30. The numerator is computed from n_success when available, otherwise from n_failed or finite summary metrics. NJ patristic RMSE/Spearman compare pairwise leaf-to-leaf patristic distances between the NJ tree reconstructed from the completed matrix and the reference ML tree. No fixed-ML-topology fitted-patristic values are used as fallback in these main columns. edge BL RMSE is taken from the ML-topology branch-length comparison. N/A indicates that the corresponding summary file or metric column was not available.}
\end{sidewaystable}



"""


# =========================
# 2. Parser
# =========================

def extract_body_rows(table_tex: str) -> list[str]:
    """
    Extract only the data body from a LaTeX table.

    Assumes the table has a header line containing:
        $p$ & Method
    and that the data body starts after the next \\midrule
    and ends before \\bottomrule.
    """
    lines = table_tex.splitlines()

    header_idx = None
    for i, line in enumerate(lines):
        if re.search(r"\$p\$\s*&\s*Method", line):
            header_idx = i
            break

    if header_idx is None:
        raise ValueError("Could not find the table header line containing '$p$ & Method'.")

    start_idx = None
    for i in range(header_idx + 1, len(lines)):
        if r"\midrule" in lines[i]:
            start_idx = i + 1
            break

    if start_idx is None:
        raise ValueError("Could not find \\midrule after the header.")

    end_idx = None
    for i in range(start_idx, len(lines)):
        if r"\bottomrule" in lines[i]:
            end_idx = i
            break

    if end_idx is None:
        raise ValueError("Could not find \\bottomrule.")

    rows = []
    for line in lines[start_idx:end_idx]:
        line = line.rstrip()
        if line.strip():
            rows.append(line)

    return rows


def add_dataset_column(rows: list[str], dataset_label: str) -> list[str]:
    """
    Add a Dataset column to each data row.

    Original rows look like:
        \\multirow{6}{*}{30\\%} & Hyb-Adam-UM & ...
        & MW$^\\star$-proj & ...

    New rows become:
        Cerc. & \\multirow{6}{*}{30\\%} & Hyb-Adam-UM & ...
        Cerc. & & MW$^\\star$-proj & ...
    """
    out = []

    for line in rows:
        stripped = line.strip()

        # Keep separators unchanged.
        if (
            stripped.startswith(r"\midrule")
            or stripped.startswith(r"\addlinespace")
            or stripped.startswith(r"\cmidrule")
            or stripped.startswith(r"\toprule")
            or stripped.startswith(r"\bottomrule")
            or stripped.startswith("%")
        ):
            out.append(line)
            continue

        # Prefix only real table rows.
        if "&" in stripped:
            out.append(f"{dataset_label} & {stripped}")
        else:
            out.append(line)

    return out


# =========================
# 3. Build combined table
# =========================

cerc_rows = add_dataset_column(
    extract_body_rows(TABLE_CERCOPITHECIDAE),
    dataset_label="Cerc."
)

het_rows = add_dataset_column(
    extract_body_rows(TABLE_HETEROGENEOUS),
    dataset_label="Het."
)

combined_latex = dedent(r"""
% Requires: \usepackage{booktabs,multirow,rotating,graphicx}
% Recommended macro: \newcommand{\best}[1]{\textbf{#1}}

\begin{sidewaystable}[p]
\centering
\caption{Combined matrix-, topology-, and ML-branch-length performance on the two
$15\times15$ mtDNA benchmarks. The Cercopithecidae dataset contains closely
related Old World monkeys, whereas the heterogeneous primate dataset spans
hominoids, Old World monkeys, New World monkeys, tarsiers, and strepsirrhines.
Matrix errors are computed only on artificially hidden entries. RMSE/MAE,
NJ-vs-ML patristic RMSE, and branch-length RMSE are reported as $\times 10^{-2}$.
Lower is better for RMSE/MAE, runtime, RF$_{\mathrm{norm}}$, NJ-vs-ML patristic
RMSE, and branch-length RMSE; higher is better for correlations. Best values
within each dataset and missingness level are in bold.}
\label{tab:combined_two_datasets}
\tiny
\setlength{\tabcolsep}{1.35pt}
\renewcommand{\arraystretch}{0.72}
\resizebox{\textwidth}{!}{%
\begin{tabular}{lllcccccccccc}
\toprule
& & & \multicolumn{5}{c}{Matrix metrics} & \multicolumn{5}{c}{Tree / branch-length metrics} \\
\cmidrule(lr){4-8}\cmidrule(lr){9-13}
Dataset & $p$ & Method
& RMSE & MAE & Pear. & Spear. & Time(s)
& RF$_{\mathrm{norm}}$ & NJ pat.\ RMSE & NJ pat.\ Spear.
& edge BL RMSE & Succ. \\
\midrule
\multicolumn{13}{l}{\textbf{Cercopithecidae dataset}} \\
\midrule
""").lstrip()

combined_latex += "\n".join(cerc_rows)

combined_latex += dedent(r"""

\midrule
\multicolumn{13}{l}{\textbf{Heterogeneous primate dataset}} \\
\midrule
""")

combined_latex += "\n".join(het_rows)

combined_latex += dedent(r"""

\bottomrule
\end{tabular}%
}
\vspace{2pt}
\parbox{\textwidth}{\scriptsize
Cerc. denotes the Cercopithecidae benchmark; Het. denotes the heterogeneous
primate benchmark. Succ. is displayed as successful/evaluated runs in the
30-run protocol. N/A indicates that the corresponding summary file or metric
column was not available, or that the method failed for that setting.}
\end{sidewaystable}
""")


# =========================
# 4. Save output
# =========================

out_path = Path("combined_two_datasets_table.tex")
out_path.write_text(combined_latex, encoding="utf-8")

print(f"Saved combined LaTeX table to: {out_path.resolve()}")
```
